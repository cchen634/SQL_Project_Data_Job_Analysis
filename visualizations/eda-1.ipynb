```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

Visualizing top-paying skills

```python
top_paying_skills = pd.read_csv('2_top_paying_skills.csv')
top_paying_skills.head()
```

```python
# Get unique skills
skills = top_paying_skills['skills'].unique()
# Create a dictionary to hold the number of time each skill appears
count = {}
for skill in skills:
    total_matches = (top_paying_skills['skills'] == skill).sum()
    count[skill] = total_matches
# Sort in descending order
sorted_count = dict(sorted(count.items(), key=lambda item:item[1], reverse=True))

plt.bar(sorted_count.keys(), sorted_count.values(), color='skyblue')
plt.xlabel('Skill')
plt.ylabel('Usage Count')
plt.title('Top-Paying Skill Frequency')
plt.xticks(rotation=45, ha='right')
plt.savefig('2_top_paying_skills.png', bbox_inches='tight')
plt.show()
```
